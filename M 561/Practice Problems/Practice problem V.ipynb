```python
# load the Chipotle dataset
url = 'https://raw.githubusercontent.com/um-perez-alvaro/Data-Science-Practice/master/Data/chipotleorders.csv'
chipo = pd.read_csv(url)
chipo.head()
```

Turn the item price into a float.

```python
chipo.item_price = chipo.item_price.replace('\$|,', '', regex=True)
```

```python
chipo.item_price = chipo.item_price.as_type('float')
```

```python
chipo.item_price = pd.to_numeric(chipo.item_price)
```
