```python
# process parts of speech

def process_pos(pos):
    if pos.startswith('J'): # adjective
        return 'a'
    elif pos.startswith('V'): # verb
        return 'v'
    elif pos.startswith('N'): 
        return 'n'
    elif pos.startswith('R'):
        return 'r'
    else:
        return 'n'
```

```python
lemmatized_words = [lemmatizer.lemmatize(word, pos=process_pos(pos))
                   for word, pos in nltk.pos_tag(words)]
lemmatized_words
```

## Yelp reviews


```python
url = 'https://raw.githubusercontent.com/um-perez-alvaro/Data-Science-Practice/master/Data/yelp.csv'
data = pd.read_csv(url)[['text', 'stars']]
data
```

Goal: predict the stars in terms of the reviews.

To simplify, we keep only the five star and one star reviews

```python
# keep 5-stars and 1-star review
data = data[data.stars.isin([1,5])].reset_index(drop=True)
```

```python
data
```

```python
data.text.iloc[0]
```

```python
text = data.text.iloc[0]
words = word_tokenize(text)

#lowercase
words = [word.lower() for word in words]

lemmatized_words = [lemmatizer.lemmatize(word, pos=process_pos(pos)) 
                    for word,pos in nltk.pos_tag(words)
                   if word not in stop_words and word not in punctuation]
" ".join(lemmatized_words)
```

```python
for i in range(len(data)):
    text = data.loc[i,'text']
    words = word_tokenize(text)
    words = [word.lower() for word in words]
    lemmatized_words = [lemmatizer.lemmatize(word, pos=process_pos(pos)) 
                    for word,pos in nltk.pos_tag(words)
                   if word not in stop_words and word not in punctuation]
    data.loc[i,'processed_text'] = " ".join(lemmatized_words)
```

```python
data
```

```python
X = data.processed_text
y = data.stars
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer # can use either
from sklearn.naive_bayes import MultinomialNB

pipe = Pipeline(steps=[
    ('vect', CountVectorizer()),
    ('clf', MultinomialNB())
])
pipe
```

```python
param_dic = {
    'vect__max_features':[2000,5000,7000,10000],
    'vect__max_df': [1.0,0.9,0.8],
    'vect__min_df': [5,20,50],
    'vect__ngram_range':[(1,1), (1,2)]
}
```

```python
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(pipe,
                   param_dic,
                   cv=5,
                   scoring='accuracy',
                   n_jobs=-1)
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
best_pipe=grid.best_estimator_

from sklearn.metrics import accuracy_score, confusion_matrix
y_test_pred = best_pipe.predict(X_test)
confusion_matrix(y_test,y_test_pred)
```

```python
accuracy_score(y_test,y_test_pred)
```

### What are the one-star words and the five-star words?

```python
vocab = best_pipe['vect'].get_feature_names_out()
vocab
```

```python
one_star_word_count = best_pipe['clf'].feature_count_[0]
five_star_word_count = best_pipe['clf'].feature_count_[1]
#dataframe
count_df = pd.DataFrame({'word':vocab,
                        '1_star':one_star_word_count,
                        '5_star':five_star_word_count}).set_index('word')

count_df
```

```python
# add 1
count_df['1_star'] += 1
count_df['5_star'] += 1
```

```python
#frequencies
count_df['1_star'] = count_df['1_star']/count_df['1_star'].sum()
count_df['5_star'] = count_df['5_star']/count_df['5_star'].sum()

#ratios
count_df['1_star_ratio'] = count_df['1_star']/count_df['5_star']
count_df['5_star_ratio'] = count_df['5_star']/count_df['1_star']
```

```python
count_df.sort_values(by='1_star_ratio', ascending=False).head(20)
```

```python
count_df.sort_values(by='5_star_ratio', ascending=False).head(20)
```
