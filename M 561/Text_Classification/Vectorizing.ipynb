```python
import pandas as pd
```

```python
# toy example: 4 documents
X_train = ['call you tonight',
          'call you a cab',
          'please call me... PLEASE',
          'he called the police']
X_train
```

We will use CountVectorizer to convert text into a matrix of word counts

```python
# import CountVectorizer
from sklearn.feature_extraction.text import CountVectorizer
vect = CountVectorizer() # with default parameters
```

```python
# 'learn the vocabulary'
vect.fit(X_train)
```

```python
vect.get_feature_names_out()
```

```python
# convert training data into a 'document-term matriox'
X_train_dtm = vect.transform(X_train)
```

```python
X_train_dtm.toarray()
```

```python
pd.DataFrame(X_train_dtm.toarray(), columns=vect.get_feature_names_out(),index=X_train)
```

```python
X_test = ["please don't call me"]
X_test_dtm = vect.transform(X_test)
pd.DataFrame(X_test_dtm.toarray(), columns=vect.get_feature_names_out(),index=X_test)
```

## Tuning the vectorizer

**stop_words:** Stop words are words like I, a, an, the, this, etc.

```python
vect = CountVectorizer(stop_words='english')
vect.fit(X_train)
vect.get_feature_names_out()
```

```python
# list of scikit leran stop words
from sklearn.feature_extraction import _stop_words
print(_stop_words.ENGLISH_STOP_WORDS)
```

**ngram_range:**

```python
vect=CountVectorizer(ngram_range = (1,3)) # 1 grams, 2 grams, and 3 grams

vect.fit(X_train)
vect.get_feature_names_out()
```

**max_df:**

```python
# ignore terns that appear in more than 50% of the documents
vect = CountVectorizer(max_df=0.5)
vect.fit(X_train)
vect.get_feature_names_out()
```
