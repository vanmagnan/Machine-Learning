## Encoding categorical features (non-numerical features)

We have two types:
- ordinal features (categories have a specific order)
    - for these features, we have ordinal features (directly below):
    - we turn categories into numbers 0,1,2,...
    - it assumes categories are ordered and 'distances' are always 1
- categorical features: no specific order
    - for these features, we turn categories into binary vectors $e_1, e_2,\ldots$

```python
data[['Sex']]
```

```python
# Ordinal encoding
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder()
encoder.fit(data[['Sex']])
encoder.transform(data[['Sex']])
```

```python
# Ordinal encoding
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder()
encoder.fit(data[['Embarked']])
encoder.transform(data[['Embarked']])
```

```python
# Ordinal encoding
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder()
encoder.fit(data[['Sex', 'Embarked']])
encoder.transform(data[['Sex', 'Embarked']])
```

```python
encoder.categories_
```

```python
# One-hot encoding

from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse_output=False) # set sparse to false if you want to visualize the encoding

encoder.fit(data[['Sex', 'Embarked']])
encoder.transform(data[['Sex', 'Embarked']])
```

```python
encoder.categories_
```

## Normalization and Standardization

(Improves the performance of Machine Learning models)

**Normalization:** scales feature values to the $[0,1]$ or $[-1,1]$ range.

   - Min Max Scaler: $\frac{\text{value}-\text{minimum value}}{\text{maximum value}-\text{minimum value}}\in [0,1]$
    
   - Max Abs Scaler: $\frac{\text{value}}{|\text{max value}|}\in [-1,1]$
   
**Standardization:** mean removal/variance scaling
    - $\frac{\text{value}-\text{mean}}{\text{variance}}$

```python
from sklearn.preprocessing import MinMaxScaler, MaxAbsScaler, StandardScaler
```

```python
scaler = MinMaxScaler()
scaler.fit(data[['Pclass', 'Age', 'Fare']])
```

```python
scaler.transform(data[['Pclass', 'Age', 'Fare']])
```

## Data Imputation: dealing with missing values

```python
from sklearn.impute import SimpleImputer
```

Missing values can be imputed with: a constant, mean value, median, mode, ...

```python
data.isna().sum()
```

```python
imputer_num = SimpleImputer(strategy = 'mean')
imputer_num.fit(data[['Age']])
imputer_num.transform(data[['Age']]) # replaces the missing values with the mean
```

```python
imputer_cat = SimpleImputer(strategy = 'most_frequent')
imputer_cat.fit(data[['Embarked']])
imputer_cat.transform(data[['Embarked']])
```

## Polynomial features / interaction features

**Polynomial Features**

ex: Pclass, Age, Fare

deg 0: 1 (constant) bias

deg 1: Pclass, Age, Fare

deg 2: Product of any two deg 1 features including squares

deg 3: All terms of degree 3

.
.
.

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
data.dropna(axis=0, subset=['Pclass','Age','Fare'], inplace = True) # this function hates NaNs
```

```python
poly = PolynomialFeatures(degree=0)
poly.fit(data[['Pclass','Age','Fare']])
poly.transform(data[['Pclass','Age','Fare']])
```

```python
poly = PolynomialFeatures(degree=1)
poly.fit(data[['Pclass','Age','Fare']])
poly.transform(data[['Pclass','Age','Fare']])
```

```python
poly = PolynomialFeatures(degree=2)
poly.fit(data[['Pclass','Age','Fare']])
poly.transform(data[['Pclass','Age','Fare']])
```

```python
# poly features names
poly.get_feature_names_out(['Pclass','Age','Fare'])
```
