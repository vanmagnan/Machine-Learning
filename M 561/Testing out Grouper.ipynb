```python
import pandas as pd
```

```python
path = 'traffic_crashes.csv'
data = pd.read_csv(path, index_col = 'CRASH_DATE', parse_dates=True)
```

```python
data['INJURIES'] = (data.INJURIES_TOTAL >= 1).astype(int)
```

What we want to replicate with a pivot table:

```python
data.groupby('INJURIES').resample('W').CRASH_RECORD_ID.count().unstack(level=0).fillna(0)
```

```python
pd.pivot_table(
    data,
    index = pd.Grouper(key=data.index, freq='W'),
    columns = 'INJURIES',
    aggfunc = 'count'
).fillna(0)
```
