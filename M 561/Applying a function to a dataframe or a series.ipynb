# Map, Apply, and Applymap

```python
import pandas as pd
import numpy as np
```

## The map method

```python
# load the Titanic dataset

url = 'https://raw.githubusercontent.com/um-perez-alvaro/Data-Science-Practice/master/Data/titanic.csv'

titanic = pd.read_csv(url)
```

```python
titanic.head()
```

```python
# we are going female to 1 and male to 0
map_dic = {'female':1, 'male':0}
titanic.Sex.map(map_dic)
```

```python
# the same, but using a function instead of a dictionary (probably more what this  is for)

def map_func(sex):
    if sex=='female':
        return 1
    else:
        return 0
```

```python
titanic.Sex.map(map_func)
```

## The apply method (with a series)

Example 1: find the length of the strings in the Name column

```python
# python len(lengths)
len('Javier')
```

```python
titanic['Name Length'] = titanic.Name.apply(len)
```

```python
titanic[['Name','Name Length']]
```

```python
# you can also use map here:

titanic.Name.map(len)
```

Example 2: extract the last name of each passenger

```python
titanic.Name[0].split(',')[0]
```

```python
def get_last_name(name):
    return name.split(',')[0]
```

```python
titanic.Name.apply(get_last_name)
```

```python
# using a lambda function
titanic.Name.apply(lambda name:name.split(',')[0])
```

## The apply method (with a dataframe)

```python
# load the alcohol consumption data
url1='https://raw.githubusercontent.com/um-perez-alvaro/Data-Science-Practice/master/Data/drinks.csv'
drinks = pd.read_csv(url1)
drinks.set_index('country',inplace=True)
```

```python
drinks.drop('continent',axis=1,inplace=True)
```

Example 1: find the maximum value in each column

```python
drinks.apply(np.max,axis=0) #axis=0 means we apply along the rows
```

```python
# alternatively, use pandas built in function
drinks.max()
```

Example 2: find the max value in each row

```python
drinks.apply(np.max,axis=1)
```

```python
drinks.max(axis=1)['USA']  # is this about wine, beer, or spirits?
```

Example 3: Find the column that has the maximum value

```python
drinks.apply(np.argmax,axis=1).map({0:'beer',1:'wine',2:'spirits',3:'pure alcohol'})
```

```python
drinks.apply(np.argmax,axis=1).map({0:'beer',1:'wine',2:'spirits',3:'pure alcohol'})[['USA','Spain']]
```
