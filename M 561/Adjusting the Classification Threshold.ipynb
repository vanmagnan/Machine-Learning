# Adjusting the Classification Threshold

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import accuracy_score, recall_score, precision_score
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
url = 'https://raw.githubusercontent.com/um-perez-alvaro/Data-Science-Practice/master/Data/pima.csv'
data = pd.read_csv(url)
```

```python
data
```

```python
# feature matrix X, target vector y

X = data.drop('label', axis=1)
y = data.label
# train test split
X_train, X_test, y_train, y_test = train_test_split(X,y)
```

```python
# fit a logistic regression model to the data
log_reg = LogisticRegression(max_iter=2000)
log_reg.fit(X_train,y_train)
```

```python
# predictions
y_test_pred = log_reg.predict(X_test)
```

```python
y_test_pred
```

```python
log_reg.predict_proba(X_test)[:,1]
```

Column 0: probability label is 0, column 1: probability that the label is 1

```python
# scikit learn uses the classification threshold of 0.5 to classify the label. We can adjust this.
```

```python
# threshold = 0.5
print(accuracy_score(y_test, y_test_pred))
print(recall_score(y_test, y_test_pred))
print(precision_score(y_test, y_test_pred))
```

```python
# predictions with a different threshold (0.7). We need to do this manually

y_test_pred = np.zeros(len(y_test))
y_test_pred
```

```python
proba = log_reg.predict_proba(X_test)[:,1]
```

```python
y_test_pred[proba>=0.7] = 1
```

```python
y_test_pred
```

```python
# threshold = 0.7
print(accuracy_score(y_test, y_test_pred))
print(recall_score(y_test, y_test_pred))
print(precision_score(y_test, y_test_pred))
```

Recall: TP, FN
Precision: TP, FP
Accuracy

Recall and Precision are opposed: increasing the threshold improve precision at the cost of recall.

```python
thresholds = np.linspace(0,1,100) # 100 evenly spaced values between 0 and 1
recalls = [] # empty list
precision = [] # empty list
# loop over thresholds
for threshold in thresholds:
    #predict
    y_test_pred = np.zeros(len(y_test))
    y_test_pred[proba>=threshold] = 1
    # precision and recall
    recalls.append(recall_score(y_test,y_test_pred))
    precision.append(precision_score(y_test,y_test_pred))
```

```python
plt.plot(thresholds,recalls, label = 'recall')
plt.plot(thresholds,precision, label = 'precision')
plt.legend()
```

```python
from sklearn.metrics import average_precision_score
y_train_pred_proba = log_reg.predict_proba(X_train)[:,-1]
average_precision_score(y_train,y_train_pred_proba)
```

## ROC curve and the AUC score

ROC (receiver operating characteristic)

```python
from sklearn.metrics import RocCurveDisplay
```

```python
RocCurveDisplay.from_estimator(log_reg, X_train, y_train)
```

In the above, AUC is the area under the curve. Values closer to 1 are preferable.

We can find this metric directly as well:

```python
# AUC metric
from sklearn.metrics import roc_auc_score
roc_auc_score(y_train, y_train_pred_proba)
```

AUC/AP Values

- 0.5 random predictions
- 0.6 poor
- 0.7 weak
- 0.8 decent
- 0.9 good
- 1.0 perfect
