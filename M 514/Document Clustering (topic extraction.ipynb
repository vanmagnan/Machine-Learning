We need a few things:

a Vectorizer (a way to represent text as numerical data).

    1) count vectorizer: a document-term matrix in which row d column t entry is the number of times term t appears in document d

```python
import numpy as np
```

```python
#4 documents
documents = ['call you tonight', 'call me a cab', 'please call me, please', 'he called the police']
documents
```

```python
#import from the scikit learn library the count-vectorizer
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer()
vectorizer.fit(documents)
vectorizer.get_feature_names()
```

```python
X_dtm = vectorizer.transform(documents)
X_dtm = X_dtm.toarray()
X_dtm
```

Now we do something more fancy-

    2) tf-idf vectorizer: tf-> term frequency x idf -> inverse document fequency
    
    tf-idf = c_dt x log[(1+n)/1+df(t)],
    where n = number of documents
    and df(t) = number of documents that contain the term t
