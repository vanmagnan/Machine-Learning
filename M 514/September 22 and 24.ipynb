LU decomposition (with partial pivoting)

1) factor PA = LU     O(n^3)

2) permute b -> Pb     free

3) solve Ly = Pb      O(n^2)

4) solve Ux = y       O(n^2)

```python
from scipy.linalg import lu_factor, lu_solve
import numpy as np
```

```python
A = [0]
```

```python
LU, p = lu_factor(A)
```

```python
b = np.array([1,1,1])
```

```python
# solve a linear system using the LU factorization and then numpy

x = lu_solve((LU,p), b)
x
```

```python
x = np.linalg.solve(A,b)
```

# A 'big' linear system

```python
n = 1000
A = np.random.randn(n,n)
b = np.random.randn(n)
```

```python
%%timeit

x = np.linalg.solve(A,b)
```

```python
%%timeit
LU,p = lu_factor(A)
x = lu_solve((LU,p), b)
```

## Solving linear systems with the same matrix, but multiple vectors b

```python
m = 200
```

```python
%%timeit
for i in range(m):
    b = np.random.randn(n)
    x = np.linalg.solve(A,b)
```

```python
%%timeit
LU,p = lu_factor(A) # this do be expensive
for i in range(m):
    b = np.random.randn(n)
    x = lu_solve((LU,p),b)
```

The second one here is a lot faster!
