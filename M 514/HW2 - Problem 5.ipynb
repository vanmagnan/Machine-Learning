```python
import numpy as np
```

```python
# generate 4 random matrices 
A = np.random.randn(1,4000) # 1x4000 random matrix
B = np.random.randn(4000,4000) # 4000x4000 random matrix
C = np.random.randn(4000,1000) # 4000x1000 random matrix
D = np.random.randn(1000,1) # 1000x1 random matrix
```

You are going to time the five different ways you can compute the product ABCD

**Way 1:** ((AB)C)D

```python
%%timeit
((A.dot(B)).dot(C)).dot(D)
```

**Way 2**: (A(BC))D

```python
%%timeit

(A.dot(B.dot(C))).dot(D)
```

**Way 3:** (AB)(CD)

```python
%%timeit
(A.dot(B)).dot(C.dot(D))
```

**Way 4:** A((BC)D)

```python
%%timeit
A.dot((B.dot(C)).dot(D))
```

**Way 5:** A(B(CD))

```python
%%timeit
A.dot(B.dot(C.dot(D)))
```
