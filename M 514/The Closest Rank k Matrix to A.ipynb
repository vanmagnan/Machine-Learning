# The closest rank k matrix to A

Matrices (from applications) are huge, where $A=m\times n$ for m,n really big. We need to store $mn$ entries!

Goal: Approximate A as a product of a $k$-row TS and $k$-column $SF$ matrix, for which we must only store $k(m+n)$ entries.

The SVD of $A$:

$A=U\Sigma V^T=[u_1|\ldots|u_m]\Sigma[v_1|\ldots|v_n]^T$. Because of the zeroes in $\Sigma$, we can write this economically as $A=[u_1|\ldots|u_r]\Sigma_r[v_1|\ldots|v_r]^T$.

So $A = \sigma_1u_1v_1^T+\ldots+\sigma_ru_rv_r^T$. We'll truncate this sum after $k$ terms to approximate $A$ as $A_k$.

That is, $A_k=[u_1|\ldots|u_k]\Sigma_k[v_1|\ldots|v_k]^T$

### Theorem (1936):

The closes rank $k$ matrix to $A$ is $A_k$.

#### What do we mean by 'closest'?

Matrix norms:

Well, vector norms satisfy the following three conditions:

   $\|v\| \geq 0$
    
   $\|v\| = 0$ only when $v = 0$
    
   $\|\alpha v\| = |\alpha | \cdot \|v\|$
    
   $\|v+w\| \geq \|v\| +\|w\|$
   


For an $m\times n$ matrix $A$, we want our norm to satisfy the following:

   $\|A\| \geq 0$
    
   $\|A\| = 0$ only when $A = 0$
    
   $\|\alpha A\| = |\alpha | \cdot \|A\|$
    
   $\|A+B\| \geq \|A\| +\|B\|$
   
   

##### Examples

Frobenius norm: $\|A\|_F = \sqrt{\sum a_{ij}}$

Induced-norms: If $\|\cdot\|$ is a vector norm, then $\|A\|$ is defined as:
$$
\text{max}_{\|x\|=1}\|Ax\|
$$

In the case of the 2-norm:

$$
\text{max}_{\|x\|_2=1}\|Ax\|_2
$$

Multiply all vectors on the unit circle by $A$. What you get is an ellipse. The farthest-away point of this ellipse has norm $\|A\|_2$. This value will be equal to $\sigma_1$, the largest singular value of $A$!

The 1-norm is given by the maximum of the absolute value column-sums

The $\infty$-norm is the maximum of the absolute value row-sums

Now that we have norms, we can talk about distances!

### Theorem:

If $B$ is a rank $k$ matrix, $\|A-B\|_{2}\geq \|A-A_k\|_2 = \sigma_{k+1}$

**Proof:**

For matrix $A$ with SVD $U\Sigma V^T$, we have $A_k=[u_1|\ldots|u_k]\Sigma_k[v_1|\ldots|v_k]^T$.

$\|A-A_k\|_2 = \|U(\Sigma_r-\Sigma_k)V^T\|_2=\|U(\Sigma_{(k+1):r})V^T$. The numbers $\sigma_{k+1},\ldots,\sigma{r}$ are the singular values of $A-A_k$. Thus its 2-norm is $\sigma_{k+1}$.

Now, let $B$ be any rank $k$ matrix.

Consider $N(B)$ and span$(v_1,\ldots,v_{k+1})$, where the vectors $v_i$ come from the SVD of $B$. By orthogonality of these vectors, we know the dimension of their span is $k+1$ and that the dimension of $N(B)$ is equal to $n-k$. Thus dim $N(B)+dim span$({v_1,\ldots,v_{k+1}) = n+1>n$, so these two subspaces share (at least) a line.

Now, $\|A-B\|_2$ is equal to the max among all unit vectors of $\|(A-B)x\|_2$. For any unit vector $x$, we have $\|(A-B)x\|_2\leq \|A-B\|_2$. Our goal now then is to find a unit vector such that $\|(A-B)x\|_2=\sigma_{k+1}$.

Choose $x\in N(B)\cap$ span$( v_1,\ldots,v_{k+1})$. $Bx = 0$ and $x=\sum c_iv_i$, for $1\leq i\leq k+1$. Thus $x=Vc$ for some vector $c$.

Then we have $\|(A-B)x\|^2 = \|Ax-Bx\| = \|Ax\|^2 = \|U\Sigma V^T x\|^2 = \|\Sigma V^Tx\|^2 = \|\Sigma V^TVx\|^2 = \|\Sigma c \|^2$

By definition, $\|\Sigma c\|^2 = \sum_{i=1}^{k+1}\sigma_i^2 c_i^2 \geq \sum_{i=1}^{k+1}\sigma_{k+1}^2 c_{i}^2=\sigma_{k+1}^2\sum_{i=1}^{k+1} c_i^2=\sigma_{k+1}^2$.
