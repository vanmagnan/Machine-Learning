# Problem 1: Book Recommendations

The [Book-Crossings dataset](http://www2.informatik.uni-freiburg.de/~cziegler/BX/) is a book ratings dataset compiled by Cai-Nicolas Ziegler from the [Book-Crossing community](https://www.bookcrossing.com). 
It contains 1.1 million ratings of 270,000 books by 90,000 users. 
The ratings are on a scale from 1 to 10.

Click [here](http://www2.informatik.uni-freiburg.de/~cziegler/BX/BX-CSV-Dump.zip) to download the dataset.
The data consists of three files: ratings (BX-Book-Ratings.csv), books info (BX-Books.csv), and users info (BX-Users.csv)

**Ratings data**

The ratings data set provides a list of ratings that users have given to books. It includes 1,149,780 records and 3 fields: userID, ISBN, and bookRating.

## Part 4

Use

- (a) Jaccard similarity $$ J(A,B) = \frac{|A\cap B|}{|A\cup B|} $$ 
- (b) serendipity similarity $$S(A,B) = \frac{|A\cap B|}{|B|}$$
- (c) overlap similarity $$O(A,B) = \frac{|A\cap B|}{\min\{|A|,|B|\}}$$
- (d) correlation similarity, and
- (e) cosine similarity

to find the top-10 book recomendations

```python
book_sets = good_ratings.groupby('ISBN')['User-ID'].apply(set)

book_sets
```

```python
# the ratings matrix
ratings_matrix = good_ratings.pivot('User-ID','ISBN')['Book-Rating']
ratings_matrix
```

```python
def Jaccard_sim(set_B):
    set_A = book_sets[book]
    return len(set_A.intersection(set_B))/len(set_A.union(set_B))

def Serendipity_sim(set_B):
    set_A = book_sets[book]
    return len(set_A.intersection(set_B))/len(set_B)

def Overlap_sim(set_B):
    set_A = book_sets[book]
    return len(set_A.intersection(set_B))/np.min([len(set_A),len(set_B)])


def cosine_sim(col_B):
    col_A = ratings_matrix.loc[:,book]
    # similarity between col_A and col_B
    norm_A = np.sqrt(np.sum(col_A**2))
    norm_B = np.sqrt(np.sum(col_B**2))
    return np.sum(col_A*col_B)/(norm_A*norm_B)

def correlation_sim(col_B):
    col_A = ratings_matrix.loc[:,book]
    
    # means
    mean_A = np.mean(col_A)
    mean_B = np.mean(col_B)
    # norms
    norm_A = np.sqrt(np.sum((col_A-mean_A)**2))
    norm_B = np.sqrt(np.sum((col_B-mean_B)**2))
    # dot product
    dot = np.sum((col_A-mean_A)*(col_B-mean_B))
    return dot/(norm_A*norm_B)
```

## Jaccard Similarity

```python
top_ten = book_sets.apply(Jaccard_sim).sort_values(ascending=False).head(10)
top_ten
```

```python
Jaccard_ten = good_books[good_books['ISBN'].isin(top_ten.index)]
Jaccard_ten
```

## Serendipity Similarity

```python
top_ten = book_sets.apply(Serendipity_sim).sort_values(ascending=False).head(10)
```

```python
Serendipity_ten = good_books[good_books['ISBN'].isin(top_ten.index)]
Serendipity_ten
```

## Overlap Similarity

```python
top_ten = book_sets.apply(Overlap_sim).sort_values(ascending=False).head(10)
```

```python
Overlap_ten = good_books[good_books['ISBN'].isin(top_ten.index)]
Overlap_ten
```

## Correlation Similarity

```python
top_ten = ratings_matrix.apply(correlation_sim).sort_values(ascending=False).head(10)
```

```python
Correlation_ten = good_books[good_books['ISBN'].isin(top_ten.index)]
Correlation_ten
```

## Cosine Similarity

```python
top_ten = ratings_matrix.apply(cosine_sim).sort_values(ascending=False).head(10)
```

```python
Cosine_ten = good_books[good_books['ISBN'].isin(top_ten.index)]
Cosine_ten
```
