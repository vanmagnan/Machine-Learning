# Gradient Descent With Momentum

First, Gradient Descent:

$$ d_k = -\nabla f(x_k), x_{k+1} = x_k - sd_k $$


Now, Gradient Descent with Momentum:

$$d_k = -\nabla f(x_k) + \beta d_{k-1}, x_{k+1} = x_k - sd_k$$

Here, $\beta$ is our momenutum coefficient

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from mpl_toolkits import mplot3d
```

```python
# example 1
def f(x,y):
    return 0.1*x**2+12*y**2
```

```python
# gradient
def grad_f(x,y):
    return np.array([0.2*x,24*y])
```

```python
# plot f
x = np.linspace(-10,10,1000)
y = np.linspace(-2,2,1000)
X,Y = np.meshgrid(x,y)
Z = f(X,Y)
```

```python
fig = plt.figure()
ax = plt.axes(projection='3d')
ax.plot_surface(X,Y,Z,cmap='seismic')
```

```python
# contour plot
plt.contour(X,Y,Z, cmap = 'coolwarm', levels=50)
```

```python
# plot gradient descent path
plt.contour(X,Y,Z,levels=50,cmap='coolwarm')

# learning rate
s = 0.08

# starting point
x = np.array([9,1])

plt.scatter(x[0],x[1],color='blue')

epochs = 200
for epoch in range(epochs):
    gradient = grad_f(x[0],x[1])
    new_x = x - s*gradient
    plt.scatter(new_x[0],new_x[1], color='blue')
    plt.plot([x[0],new_x[0]],[x[1],new_x[1]],'--', color='red')
    x=new_x
```

```python
# plot gradient descent path with momentum 
plt.contour(X,Y,Z,levels=50,cmap='coolwarm')

# learning rate
s = 0.08
b = 0.1

# starting point
x = np.array([9,1])

plt.scatter(x[0],x[1],color='blue')

# initialize directions
d_current = np.zeros(2)
d_previous = np.zeros(2)

epochs = 200
for epoch in range(epochs):
    gradient = grad_f(x[0],x[1]) 
    d_current = gradient + b*d_previous
    new_x = x - s*d_current
    plt.scatter(new_x[0],new_x[1], color='blue')
    plt.plot([x[0],new_x[0]],[x[1],new_x[1]],'--', color='red')
    x=new_x
    d_previous = d_current
```

# Binary classification problem

Binary: our problem has two classes (e.g. spam/ham, dog/cat, malignant/benign, etc)

class 1: 0

class 2: 1

goal: predict the class 0/1

example: problem in which x is the feature and y is the class (0/1 valued)

one idea: perform linear regression (hw3)

we can fit a line $y=\theta_0+\theta_1x$ to the data: if $y\geq 0.5$, then the prediction is 1. Otherwise, the prediction is 0.

However, this doesn't work very well. We are better served to use logistic regression in this instance.

## Logistic Regression:

$y=\sigma(\theta_0+\theta_1 x)$, where $\sigma$ is the sigmoid functionl again, if $y\geq 0.5$ we predict 1 and we predict 0 otherwise.

In General:

The sigmoid function $\sigma(t):\mathcal{R}\to[0,1]$ is given by 

$$\sigma(t) = \frac{1}{1+e^{-t}}$$

So, here, we have a prediction that looks like $\sigma(\theta_0+\theta_1x_1+\ldots+\theta_nx_n)$

To find the parameters, we need to minimize something:

$$\min_\theta E(\theta) = -(\sum_{i=1}^m(y_i\log p_i)+ (1-y_i)\log(1-p_i))$$

Why are we minimizing this function? Let's see what happens in the case of a wrong prediction:

if $y_i=0,p_i\approx 1, -(1-y_i)\log(1-p_i)$ is very large.

if $y_i=1,p_i\approx 0, -y_i\log p_i$ is very large.

However, in the case of a correct prediction, both terms are close to zero.

Furthermore, $E(\theta)$ is a convex function. That is, it has the property where a local minimum is also a global minimum.

Thus, to minimize $E(\theta)$, we need to solve $\nabla E = 0$

In HW 4, we'll show

$$ \nabla E = X^T(\sigma(X\theta) - y) $$

Here, we think of $\sigma(X\theta)$ as $\sigma$ being applied componentwise to the entries of $X\theta$.

Bad news: no closed-form solution of $X^T(\sigma(X\theta)-y)$
