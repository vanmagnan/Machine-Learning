```python
import numpy as np
from scipy.signal import correlate2d
```

```python
#define three matrices

X1 = np.random.choice(5, size=(3,2))
X2 = np.random.choice(5, size=(3,2))
X3 = np.random.choice(5, size=(3,2))
```

```python
# three distinct kernels

k1 = np.random.choice(3, size=(2,2))
k2 = np.random.choice(3, size=(2,2))
k3 = np.random.choice(3, size=(2,2))
```

```python
X = np.array([X1,X2,X3])
X
```

```python
K = np.array([k1,k2,k3])
K
```

```python
correlate2d(X,K,mode='valid')
```
