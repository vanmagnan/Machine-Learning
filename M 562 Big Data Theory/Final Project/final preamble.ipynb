# The Mapper Algorithm



Mapper, created by Carlsson, Memoli, and Singh in 2007, is a "...computational method for extracting simple descriptions of high dimensional data sets in the form of simplicial complexes." We present first an outline of how the algorithm functions, then implement the algorithm. We finish with applications of mapper in clustering data that k-means has trouble with.

### A Brief Overview

Mapper belongs to the broader family of Topological Data Analysis (TDA), which will often involve the creation of simplicial complexes from a data set. We will be focusing on low-dimensional simplicial complexes, which are graphs.

Here's a summary of what our implementation of the algorithm does:

1. Start with data in high dimensional space.

2. Project the data into a lower-dimensional space. 

3. Construct a cover of the projected space.

4. Create overlapping bins of data from preimages of sets in the cover.

5. Within each bin, cluster the data.

6. For each cluster in each bin, create a node in a graph.

7. Connect the nodes that share data points.

8. Return graph.


We'll now slightly expand upon each of these points:

### Expanded Overview

1. We often come across data in high dimensions. High-dimensional data is notoriously difficult to wrangle, and projecting into a lower dimensional space (often using PCA or similar tools) is a common practice. For us, that will be two-dimensional data.

2. The projection here, referred to as our 'lens,' is a function one chooses. In implementing Mapper, you may see very different results based on the selection of lens. We will be looking at the projection (x,y) -> (y) in our implementation.

3. Open covers abound in topology. In our implementation, we'll be dividing the interval of all y-values our data takes on into overlapping sub-intervals.

4. These bins of data overlap because our intervals overlap. The algorithm will not work without these overlaps!

5. You can choose any clustering algorithm you want at this point, assuming you are consistent in your choice. Here are two things to note at this point: first, we are clustering the data in its original high-dimensional space, not in the projected space. Second, in our example, slicing our data into these intervals can help us avoid some of the features that k-means has trouble with, as long as the slices are smaller than the resolution of these features (eg spirals).

6. A graph is a simplicial complex, so topology is lurking at this step.

7. This is where our overlapping bins is especially relevant. If no bins share data points, then no clusters will either!

8. The graph we return can then be analyzed in a few different ways. In our context, we are finding connected components and clustering our original data based off the connected component their clusters fall in.
