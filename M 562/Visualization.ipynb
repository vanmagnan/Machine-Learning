```python
import pandas as pd
```

```python
path = 'traffic_crashes.csv'
data = pd.read_csv(path, index_col = 'CRASH_DATE', parse_dates=True)
```

```python
data
```

```python
# filter out years 2013 - 2017
data = data.loc[data.index.year>=2018]
```

```python
data.head()
```

```python
data.resample('W').CRASH_RECORD_ID.count().plot()
```

```python
# how has the injury rate changed over time

data['INJURIES'] = (data.INJURIES_TOTAL >= 1).astype(int)
```

```python
data.INJURIES.mean()
```

```python
data.groupby('INJURIES').resample('W').CRASH_RECORD_ID.count().unstack(level=0)
```

```python
# the same, using pivot_table
data.pivot_table(values='CRASH_RECORD_ID', index = pd.Grouper(key=data.index.CRASH_DATE, freq='1W'), columns = data.index.INJURIES, aggfunc = 'count')
```

```python
# number of crashes that involve injuries

n_injuries = data.loc[data.INJURIES==1].resample('W').CRASH_RECORD_ID.count()
```

```python
n_injuries
```

```python
n_total = data.resample('W').CRASH_RECORD_ID.count()
```

```python
pct_weekly_injuries = 100*n_injuries/n_total
```

```python
pct_weekly_injuries.plot()
```

```python
# how do injuries vary with first crash type


data.FIRST_CRASH_TYPE.unique()
```

```python
table = data.pivot_table(values = 'CRASH_RECORD_ID', index = 'FIRST_CRASH_TYPE', columns = 'INJURIES', aggfunc = 'count')
```

```python
table
```

```python
table['total']= table.sum(axis=1)
```

```python
table
```

```python
table['pct_injuries'] = table[1]/table['total']
```

```python
table.pct_injuries.plot.bar()
```

```python
## are injuries more likely in different locations?

mask = (data.LONGITUDE!=0)& (data.LATITUDE!=0)& (data.INJURIES ==1)

data.loc[mask].plot.scatter(x='LONGITUDE', y ='LATITUDE', figsize=(7,7), s=0.1, alpha=0.5)
```

```python
data[mask].plot.hexbin(x='LONGITUDE', y ='LATITUDE', gridsize=20,  figsize=(7,7),cmap='YlOrBr')
```

```python
data.resample('D').CRASH_RECORD_ID.count().plot.hist(bins=20) # histogram
```

```python
#probability distribution function

data.resample('D').CRASH_RECORD_ID.count().plot.density()
```

```python
# Area plot
# weekly number of crashes
data.resample('W').CRASH_RECORD_ID.count().plot(figsize=(15,5)) 
```

```python
# weekly number of crashes by crash type

data.groupby('FIRST_CRASH_TYPE').resample('W').CRASH_RECORD_ID.count().unstack(level=0).plot(figsize=(15,7),legend=False)
```

```python
data.groupby('FIRST_CRASH_TYPE').resample('W').CRASH_RECORD_ID.count().unstack(level=0).plot.area(figsize=(15,7))
```

```python
# without the legend it makes it uglier

data.groupby('FIRST_CRASH_TYPE').resample('W').CRASH_RECORD_ID.count().unstack(level=0).plot.area(figsize=(15,7), legend = False)
```

## Advanced plotting

```python
import matplotlib.pyplot as plt
```

```python
# weekly number of crashes by injury status (0=no, 1=yes)
weekly = data.groupby('INJURIES').resample('W').CRASH_RECORD_ID.count().unstack(level=0)
weekly
```

```python
# create figure and axis
fig, axis = plt.subplots(figsize=(15,5))
# plot weekly dataframe
weekly.plot(ax=axis)
# change x-axis label
axis.set_xlabel('date', fontsize=15)
# add y-axis label
axis.set_ylabel('weekly number of crashes', fontsize=15)
# add title
axis.set_title('Number of traffic crashes per week', fontsize=20)
# modify the legend
axis.legend(labels=['no','yes'], fontsize=15, loc = 'upper left',title='injuries?', title_fontsize=18)
# add grid lines
axis.grid(True)
```

```python
# create figure and axis
fig, axis = plt.subplots(1,2,figsize=(15,5))
# plot weekly dataframe
weekly[0].plot(ax=axis[0])
weekly[1].plot(ax=axis[1],color='red')
# change x-axis labels
axis[0].set_xlabel('date', fontsize=15)
axis[1].set_xlabel('date', fontsize=15)
# set titles
axis[0].set_title('injuries = 0')
axis[1].set_title('injuries = 1')


fig.suptitle('Number of traffic crashes per week', fontsize=20)
axis[0].grid(True)
axis[1].grid(True)
```
