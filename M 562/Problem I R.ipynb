```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Problem 1: Linear regression

Your goal is to fit a linear regression model to the following two datasets. Plot the fitted line and the datapoints.

## Dataset 1: The song of insect

The following data (*from The Song of Insects by George W. Pierce, 1948*) shows the relationship between chirps (pulses of sound) per second of a cricket and the corresponding ground temperature (measured in degrees Farenheit).

```python
url = 'https://raw.githubusercontent.com/um-perez-alvaro/Data-Science-Theory/master/Data/crickets.csv' 
data = pd.read_csv(url) 
data.head(5)
```

```python
# import the linear regression model from the sci-kit learn library
from sklearn.linear_model import LinearRegression
# instantiate the model
linreg = LinearRegression()
# fit the linear regression model to the data
linreg.fit(np.asarray(data['Chirps/Second']).reshape(-1, 1), np.asarray(data['Temperature (F)']).reshape(-1, 1))
```

```python

m = 100
x_plot = np.linspace(14,20,m).reshape(100,1)
y_plot = linreg.predict(x_plot)

fig, ax = plt.subplots(figsize=(12,5)) # figure size
ax.plot(x_plot,y_plot, c='red', label='linear model')

# plot the dataset
plt.scatter(data['Chirps/Second'], data['Temperature (F)']) # plot the data
ax.set_xlabel('Temperature (F)',fontsize=20) # set x-axis label
ax.set_ylabel('y',fontsize=20) # set y-axis label
ax.set_title('The song of an insect',fontsize=20) # set title
ax.legend(fontsize=15) # plot legend

```

## Dataset 2: The rise of student loan debt in the United States

This dataset contains the total student loan debt, that is money borrowed by students to pay for college tuition, room and board, etc., held by citizens of the United States from 2006 to 2014, measured quarterly.

```python
'load the data'
url = 'https://raw.githubusercontent.com/um-perez-alvaro/Data-Science-Theory/master/Data/student%20debt.csv'
data = pd.read_csv(url)
data.head()
```

```python
# your code here
x = np.asarray(data.year).reshape(-1, 1)
y= np.array(data.debt).reshape(-1, 1)
```

```python
# instantiate the model
linreg = LinearRegression()
# fit the linear regression model to the data
linreg.fit(x,y)
```

```python


# plot the linear regression model (only for 2 dimensional data)
m = 1000
x_plot = np.linspace(2004,2021.5,m).reshape(m,1)
y_plot = linreg.predict(x_plot)

fig, ax = plt.subplots(figsize=(12,5)) # figure size
ax.plot(x_plot,y_plot, c='red', label='linear model')

# plot the dataset
ax.scatter(x,y, label='data points') # plot the data
ax.set_xlabel('Year',fontsize=20) # set x-axis label
ax.set_ylabel('Debt',fontsize=20) # set y-axis label
ax.set_title('Student Loan Debt',fontsize=20) # set title
ax.legend(fontsize=15) # plot legend

```
