```python
# set Time column as index
ufo.set_index('Time', 
                 inplace=True)
#inplace=True means what?
```

```python
# create new Location column
ufo['Location'] = ufo.City + ', ' + ufo.State 
```

```python
#  Rename the 'Colors Reported' and 'Shape Reported' columns to 'colors_reported' and 'shape_reported'

# create a dictionary with the new names
new_column_names = {'Colors Reported':'colors_reported', 'Shape Reported':'shape_reported'}

# rename columns
ufo.rename(columns=new_column_names, inplace=True)
ufo.head()
```

```python
# Remove the 'colors_reported' column

ufo.drop('colors_reported', axis=1, inplace=True) 
```

```python
# Remove rows 0 and 1



ufo.reset_index(inplace=True,
                   drop=False)

ufo.drop([0,1], axis=0, inplace=True)
```
