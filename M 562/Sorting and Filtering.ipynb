# Sorting

```python
import pandas as pd
```

```python
# load the 'imdb_1000 dataframe'
url = 'https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/imdb_1000.csv'
movies = pd.read_csv(url)
movies.head()
```

```python


movies.title.sort_values() # ; returns a Series sorted in (ascending) alphabetical order

```

```python
movies.title.sort_values(ascending=False) # by default: ascending=True
```

```python
# sort by 'star rating' and 'duration'
movies.sort_values(by=['star_rating','duration'], ascending=[True,False])
```

# Filtering

```python
# example 1: get movies that are at least 200 minutes long
movies.loc[movies.duration>=200]
```

```python
movies[movies.duration>=200]
```

```python
# why using loc method is better

# if duration >= 200, replace duration 'too long'

movies.loc[movies.duration>=200, 'duration'] = 'too long'
```

```python
movies
```

```python
# load the 'imdb_1000 dataframe'
url = 'https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/imdb_1000.csv'
movies = pd.read_csv(url)
```

```python
# try again without using loc

movies[movies.duration>=200].duration = 'too long'
```

```python
# get movies with missing 'content rating'

movies.loc[movies.content_rating.isna()]
```

```python
# get dramas that are at least 200 minutes long
movies.head()
```

```python
movies.loc[(movies.genre=='Drama') & (movies.duration>=200)]
```

```python
# get all movies with genre: Drama, Crime, or Action

movies.loc[(movies.genre=='Drama') | (movies.genre=='Crime') | (movies.genre=='Action')]
```

```python
movies.loc[movies.genre.isin(['Drama','Crime','Action'])]
```

```python
# get all dramas and crime that are at least 200 minutes long

movies.loc[movies.genre.isin(['Drama','Crime']) & (movies.duration>=200)]
```
